```python
def dls(graph, node, goal, limit, path):
    if node == goal:
        return path

    if limit <= 0:
        return None

    for neighbour in graph[node]:
        if neighbour not in path:
            result = dls(graph, neighbour, goal, limit-1, path+[neighbour])
            if result:
                return result

    return None


# -------- MAIN --------
graph = {}

n = int(input("Enter number of nodes: "))

for i in range(n):
    node = input("\nEnter node name: ")
    neighbours = input("Enter neighbours separated by space: ").split()
    graph[node] = neighbours

start = input("\nEnter start node: ")
goal = input("Enter goal node: ")
limit = int(input("Enter depth limit: "))

result = dls(graph, start, goal, limit, [start])

if result:
    print("\nDLS Path Found:")
    print(" -> ".join(result))
    print("Total Steps:", len(result)-1)
else:
    print("No Path Found within given depth limit")
```
