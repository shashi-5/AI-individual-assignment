```python
def dls(graph, node, goal, limit, path):
    if node == goal:
        return path

    if limit <= 0:
        return None

    for neighbour in graph.get(node, []):   # SAFE VERSION
        if neighbour not in path:
            result = dls(graph, neighbour, goal, limit-1, path+[neighbour])
            if result:
                return result
    return None


def iddfs(graph, start, goal, max_depth):
    for depth in range(max_depth + 1):
        result = dls(graph, start, goal, depth, [start])
        if result:
            return result
    return None


# -------- MAIN --------
graph = {}

n = int(input("Enter number of nodes: "))

for i in range(n):
    node = input("\nEnter node name: ")
    neighbours = input("Enter neighbours separated by space: ").split()
    graph[node] = neighbours

start = input("\nEnter start node: ")
goal = input("Enter goal node: ")
max_depth = int(input("Enter maximum depth: "))

result = iddfs(graph, start, goal, max_depth)

if result:
    print("\nIDDFS Path Found:")
    print(" -> ".join(result))
    print("Total Steps:", len(result)-1)
else:
    print("No Path Found within maximum depth")
```
